```python
# ==========================================
# STEP 1: LOAD LIBRARIES AND IMPORT DATA
# ==========================================
import pandas as pd
import sqlite3
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset (Make sure your file is uploaded and named exactly 'pedestrian_data.csv')
try:
    df = pd.read_csv('/content/pedestrian_data.csv.csv')
    print("CSV file loaded successfully!")
except FileNotFoundError:
    print("ERROR: 'pedestrian_data.csv' not found. Please upload your CSV to the left sidebar and rename it.")

# Connect to a virtual SQL database and save the table
conn = sqlite3.connect('melbourne_pedestrian_analytics.db')
df.to_sql('pedestrian_traffic', conn, if_exists='replace', index=False)
print(" SQL database structured. Table 'pedestrian_traffic' is ready for querying.\n")

```

```python
# ==========================================
# STEP 2: RUN THE FIXED SQL QUERY
# ==========================================
# This version groups by location and finds the absolute highest traffic zones to ensure data displays!
fixed_query = """
SELECT
    sensor_name,
    AVG(CAST(pedestriancount AS REAL)) AS average_hourly_traffic
FROM pedestrian_traffic
WHERE sensor_name IS NOT NULL
GROUP BY sensor_name
ORDER BY average_hourly_traffic DESC
LIMIT 10;
"""

# Fetch the results from the SQL database into a DataFrame
rush_hour_results = pd.read_sql_query(fixed_query, conn)

print("--- TOP 10 BUSIEST LOCATIONS (SQL OUTPUT) ---")
print(rush_hour_results)
print("-----------------------------------------------\n")
```

```python
# ==========================================
# STEP 3: DATA VISUALIZATION (THE BAR CHART)
# ==========================================
# Check if we actually have data before trying to draw the chart
if rush_hour_results.empty:
    print(" Query returned 0 rows. Please check if 'sensor_name' or 'pedestriancount' match your file exactly.")
else:
    plt.figure(figsize=(12, 6))
    sns.set_theme(style="whitegrid")

    # Create the horizontal bar plot
    sns.barplot(
        x='average_hourly_traffic',
        y='sensor_name',
        data=rush_hour_results,
        palette='viridis',
        hue='sensor_name',
        legend=False
    )

    # Add clear chart labels and titles for business managers
    plt.title('Top 10 Busiest Melbourne CBD Pedestrian Hotspots', fontsize=14, fontweight='bold', pad=15)
    plt.xlabel('Average Hourly Pedestrian Count', fontsize=12, fontweight='bold')
    plt.ylabel('Sensor / Location Name', fontsize=12, fontweight='bold')

    plt.tight_layout()

    # Save the plot directly to your Colab folder as an image for GitHub
    plt.savefig('dashboard_preview.png', dpi=300)
    print("Chart generated and auto-saved as 'dashboard_preview.png'!")

    plt.show()
```
